# Hafta 3 – EDA: Türkçe Toplantı Test Seti

```
import pandas as pd, json, glob, os, re
import matplotlib.pyplot as plt
m = pd.read_csv('../data/manifest.csv'); m
```

```
rows=[]
for f in glob.glob('../data/*/ref/*.txt'):
    t=open(f,encoding='utf-8').read(); i=os.path.basename(f)[:-4]
    g=json.load(open(f.replace('ref','tasks').replace('.txt','.json'),encoding='utf-8'))
    rows.append({'id':i,'kelime':len(t.split()),'cümle':len(re.split(r'[.!?]',t)),'görev':len(g),
                 'atanmadı':sum(x['assignee']=='atanmadı' or x['assignee'] is None for x in g),
                 'tarihsiz':sum(x['due'] is None for x in g)})
df=pd.DataFrame(rows); df
```

```
df.set_index('id')[['kelime','görev']].plot.bar(); plt.title('Toplantı başına kelime / görev'); plt.show()
```

```
# Code-switching: sözlük terimlerinin referans transkriptlerdeki sıklığı
from collections import Counter
c=Counter()
for _,r in m.iterrows():
    for f in glob.glob(f'../data/*/ref/{r.id}.txt'):
        t=open(f,encoding='utf-8').read().lower()
        for g in r.glossary.split(';'): c[g]+=t.count(g.lower())
pd.Series(c).sort_values().plot.barh(); plt.title('Terim sıklığı'); plt.show()
```
